```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
df = np.round(pd.read_csv('50_Startups.csv')[['R&D Spend','Administration','Marketing Spend','Profit']]/10000)
```

```python
np.random.seed(9)
```

```python
df = df.sample(5)
```

```python
df
```

```python
df = df.iloc[:, 0: -1]
```

```python
df.iloc[1,0] = np.nan
df.iloc[3,1] = np.nan
df.iloc[-1,-1] = np.nan

# we are working with the original dataset thats the reason we are getting this warning
```

```python
df.head()
```

```python
# iterative imputer is a technique in wich we are filling the data repeatedly until we get proper values in place of null values
```

```python
# so the 1st iteration i will be doing is mean imputation
```

```python
# brutefore approach
```

```python
df0 = pd.DataFrame()
```

```python
df0['R&D Spend'] = df['R&D Spend'].fillna(df['R&D Spend'].mean())
df0
```

```python
df0['Administration'] = df['Administration'].fillna(df['Administration'].mean())
df0
```

```python
df0['Marketing Spend'] = df['Marketing Spend'].fillna(df['Marketing Spend'].mean())
df0
```
